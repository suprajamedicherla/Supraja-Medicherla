**Barplots**

```python
plt.figure(figsize=(10,10))
sns.barplot(x='avg_humidity_relative_2m_pct', y='max_humidity_specific_2m_gpkg', data=weathersource)
```

```python
county_codes = [19061, 27153, 17111]
filtered_data = weathersource[weathersource['county_fips_code'].isin(county_codes)]

sns.histplot(x="avg_humidity_specific_2m_gpkg", hue="county_fips_code", data=filtered_data, multiple="stack")

plt.xlabel("Average Humidity Specific (2m) g/kg")
plt.ylabel("Frequency")
plt.title("Histogram of Average Humidity Specific by County")

plt.show()
```
